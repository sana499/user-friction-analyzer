```python
# 3. Encode categorical variables and save the encoders!
encoders = {}
for col in ['Subscription Type', 'Contract Length', 'Churn']:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    encoders[col] = le

X = df.drop(columns=['Churn'])
y = df['Churn']
```

```python
# 4. Scale the numerical data and save the scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("2. Training Custom Lightweight Model...")
rf_model = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
rf_model.fit(X_scaled, y)

print("3. Exporting Assets for Deployment...")
joblib.dump(rf_model, 'churn_model.pkl')
joblib.dump(scaler, 'scaler.pkl')
joblib.dump(encoders, 'encoders.pkl')

print("✅ Success! You now have churn_model.pkl, scaler.pkl, and encoders.pkl")
```
